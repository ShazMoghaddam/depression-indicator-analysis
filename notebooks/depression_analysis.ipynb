# Who gets told they have depression?

The calculations live in [`findings/analysis.py`](../findings/analysis.py) and the charts in [`findings/charts.py`](../findings/charts.py), drawn with [bezel-charts](https://pypi.org/project/bezel-charts/). This notebook calls the same functions, so its numbers always match the report.

## 2. Is the 2017 jump real?

Four checks: how many groups rose, how many people answered, the margin of error, and the scaled-up count.

```python
an.changes(2016, 2017).drop(index="Total").sort_values().round(1).to_frame("change, points")
```

```python
pd.DataFrame({"people answering": an.survey_size(), "margin, ± points": an.margin().round(2), "adults with depression": an.weighted_adults()})
```

## What this data can't tell you

It counts diagnoses, not depression, and the 2017 change can't be explained from the data alone. The [findings report](../findings/README.md) sets out the limits in full.
